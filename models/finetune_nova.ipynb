```
import os
from ultralytics import YOLO
model = YOLO('/content/nova_dataset/best.pt')
model.train(data='/content/nova_dataset/data.yaml', epochs=25, imgsz=640, batch=16, name='nova_ft', exist_ok=True)
best = os.path.join(model.trainer.save_dir, 'weights', 'best.pt')
print('Saved to', best)
```

```
# Accuracy per object (mAP50: 1.0 = perfect), then download
YOLO(best).val(data='/content/nova_dataset/data.yaml')
files.download(best)
```
