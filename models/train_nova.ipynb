```
from ultralytics import YOLO
model = YOLO('yolov8s.pt')  # start from the stock model, so it already knows what objects look like
model.train(data='/content/nova_dataset/data.yaml', epochs=60, imgsz=640, batch=16, patience=15, name='nova')
```

```
# Accuracy per object (mAP50: 1.0 = perfect)
metrics = YOLO('runs/detect/nova/weights/best.pt').val(data='/content/nova_dataset/data.yaml')
files.download('runs/detect/nova/weights/best.pt')
```
